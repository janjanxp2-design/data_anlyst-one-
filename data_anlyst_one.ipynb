```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("airline_passenger_satisfaction.csv")
df
```

```python
df.info()
```

```python
df["Gender"].value_counts()
```

```python
gender_counts=df["Gender"].value_counts()
plt.figure(figsize=(5,3))
count=plt.bar(gender_counts.index,gender_counts.values,color=['navy','skyblue'])
plt.ylim(0,max(gender_counts)*1.2)
plt.bar_label(count,padding=2)
plt.title("Comparison of male & female")
plt.ylabel("Count")
plt.show()
```

```python
customer_type=df["Customer Type"].value_counts()
plt.figure(figsize=(5,3))
plt.bar(customer_type.index,customer_type.values,color=["red","green"],alpha=0.6)
plt.title("Customer Type")
plt.xlabel("Counts")
plt.ylabel("Customer Type")
plt.show()
```

```python
age_counts=df["Age"].value_counts()
plt.figure(figsize=(5,3))
plt.hist(df["Age"],bins=10,color="skyblue",edgecolor="black")
plt.title("Age Graph(Histgram)")
plt.ylabel("Counts")
plt.xlabel("Age")
plt.show()
```

```python
plt.figure(figsize=(5,3))
plt.boxplot(df["Departure and Arrival Time Convenience"],vert=True)
plt.title("Time Convenience - boxplt")
plt.ylabel("Counts")
plt.grid()
plt.xlabel("Departure and Arrival Time Convenience")
plt.show()
```

```python
tot_count=df['Type of Travel'].value_counts()
plt.figure(figsize=(5,5))
plt.pie(tot_count,labels=tot_count.index,autopct="%1.1f%%", startangle=90)
plt.title("'Types of Travel'")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

# kde=True aur stat="count" se shape KDE jaisi rahegi par Y-axis par Taadad aayegi
sns.histplot(
    data=df,
    x="Age",
    hue="Satisfaction",
    kde=True,
    stat="count",
    element="poly",  # Continuous line curve ke liye
    fill=True,
    palette="Set2",
    alpha=0.4,
)

plt.grid(True, linestyle="--", alpha=0.6)
plt.title("Age Distribution by Satisfaction (Passenger Count)")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")  # Direct Taadad
plt.show()
```

```python
plt.figure(figsize=(10,8))
plt.scatter(df['In-flight Entertainment'],df['Baggage Handling'], color="purple")
plt.title('Entertainment values at flight')
plt.xlabel('Entertainments')
plt.ylabel("Density")
plt.show()
```

```python
import matplotlib.pyplot as plt
grouped = (
    df.groupby("Age")[["Cleanliness", "Food and Drink", "In-flight Service"]].mean().reset_index())
plt.figure(figsize=(10, 5))
plt.stackplot(grouped["Age"],grouped["Cleanliness"],grouped["Food and Drink"],grouped["In-flight Service"],
    labels=["Cleanliness", "Food and Drink", "In-flight Service"],
    colors=["#66c2a5", "#fc8d62", "#8da0cb"],)
plt.title("Average Service Ratings by Age")
plt.xlabel("Age")
plt.ylabel("Cumulative Score")
plt.legend(loc="upper right")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
